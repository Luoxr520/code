```python
from pathlib import Path
import shutil
import subprocess
import sys
import importlib.util
import json
import random

CURRENT_DIR = Path.cwd()

print("当前工作目录：", CURRENT_DIR)

if CURRENT_DIR.name == "scripts":
    ROOT_DIR = CURRENT_DIR.parent
elif (CURRENT_DIR / "scripts").exists():
    ROOT_DIR = CURRENT_DIR
else:
    ROOT_DIR = Path(r"d:\code\yolo")

print("项目根目录：", ROOT_DIR)
```

```python
DATASETS_DIR = ROOT_DIR / "datasets"
RAW_IMAGES_DIR = DATASETS_DIR / "raw_images"
UNLABELED_DIR = RAW_IMAGES_DIR / "unlabeled"

LABELED_DATASET_DIR = DATASETS_DIR / "labeled_yolo_v1"

LABELED_IMAGES_DIR = LABELED_DATASET_DIR / "images"
LABELED_LABELS_DIR = LABELED_DATASET_DIR / "labels"

TRAIN_IMAGES_DIR = LABELED_IMAGES_DIR / "train"
VAL_IMAGES_DIR = LABELED_IMAGES_DIR / "val"

TRAIN_LABELS_DIR = LABELED_LABELS_DIR / "train"
VAL_LABELS_DIR = LABELED_LABELS_DIR / "val"

OUTPUTS_DIR = ROOT_DIR / "outputs"
LABELING_CHECK_DIR = OUTPUTS_DIR / "labeling_check"
PREVIEW_DIR = LABELING_CHECK_DIR / "preview"

MODELS_DIR = ROOT_DIR / "models"

for path in [
    UNLABELED_DIR,
    LABELED_DATASET_DIR,
    TRAIN_IMAGES_DIR,
    VAL_IMAGES_DIR,
    TRAIN_LABELS_DIR,
    VAL_LABELS_DIR,
    LABELING_CHECK_DIR,
    PREVIEW_DIR,
    MODELS_DIR,
]:
    path.mkdir(parents=True, exist_ok=True)

print("原始未标注图片目录：", UNLABELED_DIR)
print("复核后 YOLO 数据集目录：", LABELED_DATASET_DIR)
print("模型目录：", MODELS_DIR)
```

```python
USE_COCO128_AS_TEMP_RAW_IMAGES = True
MAX_TEMP_IMAGES = 30

if USE_COCO128_AS_TEMP_RAW_IMAGES:
    COCO128_IMAGE_DIR = DATASETS_DIR / "coco128" / "images" / "train2017"

    if not COCO128_IMAGE_DIR.exists():
        raise FileNotFoundError(f"未找到 COCO128 原始图片目录：{COCO128_IMAGE_DIR}")

    source_images = sorted(
        list(COCO128_IMAGE_DIR.glob("*.jpg")) +
        list(COCO128_IMAGE_DIR.glob("*.jpeg")) +
        list(COCO128_IMAGE_DIR.glob("*.png"))
    )

    selected_images = source_images[:MAX_TEMP_IMAGES]

    for src in selected_images:
        dst = UNLABELED_DIR / src.name
        if not dst.exists():
            shutil.copy2(src, dst)

    print("已复制测试原始图片数量：", len(selected_images))
else:
    print("跳过测试图片复制，请手动把原始图片放入：", UNLABELED_DIR)

print("当前未标注图片数量：", len(list(UNLABELED_DIR.glob('*.*'))))
```

```python
image_paths = sorted(
    list(UNLABELED_DIR.glob("*.jpg")) +
    list(UNLABELED_DIR.glob("*.jpeg")) +
    list(UNLABELED_DIR.glob("*.png")) +
    list(UNLABELED_DIR.glob("*.bmp"))
)

print("待标注图片数量：", len(image_paths))

for path in image_paths[:10]:
    print(path.name)

if len(image_paths) == 0:
    raise RuntimeError(f"请先把原始图片放入：{UNLABELED_DIR}")
```

```python
PROJECT_CLASSES = [
    "person",
]

CLASSES_TXT_PATH = LABELED_DATASET_DIR / "classes.txt"

with open(CLASSES_TXT_PATH, "w", encoding="utf-8") as f:
    for name in PROJECT_CLASSES:
        f.write(name + "\n")

print("classes.txt 已生成：", CLASSES_TXT_PATH)
print(PROJECT_CLASSES)
```

```python
def launch_ser():
    exe_path = shutil.which("ser")

    if exe_path is None:
        print("没有找到 ser 启动命令。")
        print("如果已安装，尝试重启 Kernel；或者先使用 SelfEvolvingRecognition。")
        return None

    print("启动 SelfEvolvingRecognition：", exe_path)
    print("图片目录：", UNLABELED_DIR)
    print("输出目录：", LABELED_DATASET_DIR)
    print("类别文件：", CLASSES_TXT_PATH)

    process = subprocess.Popen([
        exe_path,
        "--filename",
        str(UNLABELED_DIR),
        "--output",
        str(LABELED_DATASET_DIR),
        "--labels",
        str(CLASSES_TXT_PATH),
        "--autosave"
    ])

    return process


# 如果你安装并想使用 SelfEvolvingRecognition，再运行这一行
ser_process = launch_ser()
```
